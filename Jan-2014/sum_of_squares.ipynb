# Jane Street Puzzle (Jan 2014) - Sum of Squares

Goal: fill a 5x5 grid with digits 0-9 so that every row/column sum of squares is unique and matches the provided clues.

Notebook layout:

- Imports and helpers.
- solve_grid() builds/validates a grid via brute-force + pruning.
- erify(g) checks a candidate grid against the constraints.
- Bottom cells run the solver and display the result.

Usage: run all cells; the solver prints/returns the grid.


```python
import numpy as np
from functools import lru_cache
```

```python
def solve_grid():
    # Build empty 5x5 grid
    g = [[0]*5 for _ in range(5)]

    # Column 4 (index 3) all 9s
    for r in range(5):
        g[r][3] = 9

    # Column 3 (index 2): last three rows 8,8,8 for Ã·8
    g[2][2] = g[3][2] = g[4][2] = 8
    g[0][2] = g[1][2] = 9  # free to maximize

    # Column 5 (index 4) bottom 0 (Ã·10), also satisfies row5 Ã·5
    g[4][4] = 0

    # Column 2 (index 1): top must be 8 so that 8,9,9,9,9 is Ã·7
    g[0][1] = 8
    for r in range(1,5):
        g[r][1] = 9

    # Column 1 (index 0): 9,9,7,9,8 gives even and sumâ‰¡0 (mod3) -> Ã·6
    g[0][0], g[1][0], g[2][0], g[3][0], g[4][0] = 9, 9, 7, 9, 8

    # Row 4 (index 3): last two must be multiple of 4 with col4=9 -> use 96
    g[3][4] = 6

    # Row 2 (index 1) must be even -> make last digit 8 (max)
    g[1][4] = 8

    print(g)
    # Fill remaining maximally with 9s
    for r in range(5):
        for c in range(5):
            if g[r][c] == 0 and not (r == 4 and c == 4):
                g[r][c] = 9

    return g
    
```

```python
def verify(g):
    row_divs = [1,2,3,4,5]
    col_divs = [6,7,8,9,10]
    def num(seq): return int(''.join(map(str, seq)))
    # rows
    for r,d in enumerate(row_divs):
        assert num(g[r]) % d == 0
    # cols
    for c,d in enumerate(col_divs):
        col = [g[r][c] for r in range(5)]
        assert num(col) % d == 0
    return True
```

```python
grid = solve_grid()
assert verify(grid)
best_sum = sum(sum(r) for r in grid)

print("Best sum:", best_sum)
for row in grid: print(*row)
submission = (best_sum,) + tuple(v for row in grid for v in row)
print("\nSubmission tuple:")
print(submission)
```
